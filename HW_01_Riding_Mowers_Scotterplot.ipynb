```python
import pandas as pd
from google.colab import drive
drive.mount('/content/drive')

mowers_df = pd.read_csv('/content/drive/MyDrive/RidingMowers.csv')
mowers_df.head()
```

```python
#Create a scatterplot of Incone vs Lot Size using color or symbol
# to distinguish owners from nonowners
fig, ax = plt.subplots()

subset=mowers_df.loc[mowers_df['Ownership']=='Owner']
ax.scatter(subset.Lot_Size, subset.Income, color='blue', marker = 'o', label='Owner')

subset=mowers_df.loc[mowers_df['Ownership']=='Nonowner']
ax.scatter(subset.Lot_Size, subset.Income, color='red', marker = 'D', label='Nonowner')

plt.xlabel('Lot Size')
plt.ylabel('Income')

handles, labels = ax.get_legend_handles_labels()
ax.legend(handles, labels)
```
