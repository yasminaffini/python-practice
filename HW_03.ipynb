```python
# AN420
# Exercise 3.3 (p. 98)
#
# Import required packages
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/drive/My Drive/Colab Notebooks/LaptopSalesJanuary2008.csv')

#Grouping by the mean

avg_price_by_store = df.groupby('Store Postcode')['Retail Price'].mean().sort_values()

#Barchart

plt.figure(figsize=(12, 6))
avg_price_by_store.plot(kind='bar', color='steelblue')
plt.xlabel('Store (Postcode)')
plt.ylabel('Average Retail Price (£)')
plt.title('Average Retail Price by Store - January 2008')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
#Import required packages

import seaborn as sns

order = df.groupby('Store Postcode')['Retail Price'].median().sort_values().index
subset = df[df['Store Postcode'].isin(['N17 6QA', 'W4 3PH'])]

#Create boxplot

plt.figure(figsize=(6, 6))
sns.boxplot(data=subset, x='Store Postcode', y='Retail Price', order=['W4 3PH', 'N17 6QA'])
plt.xlabel('Store (Postcode)')
plt.ylabel('Retail Price (£)')
plt.title('Retail Price: W4 3PH vs N17 6QA')
plt.tight_layout()
plt.show()
```
